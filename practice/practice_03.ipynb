## 2 · Looping over a list

A list holds several values in order, inside square brackets. A `for` loop runs its indented body once for each item, binding the loop variable to that item.

```python
for reading in readings:
    print(reading)
```

The colon opens the body; the indented lines *are* the body. Four spaces, every time.

````python
readings = [0.41, 0.38, 0.55, 0.62, 0.29]

for reading in readings:
    print(reading)
````

The body can be more than one line. Everything indented under the colon runs once per pass.

````python
for reading in readings:
    doubled = reading * 2
    print(reading, "->", doubled)
````

## 4 · Updating a value on every pass

Set up a value **before** the loop, update it **inside** the loop, use it **after**. Watch `total` grow — the square-root loop does exactly this to `guess`.

````python
total = 0

for r in readings:
    total = total + r
    print("added", r, "-> total is now", round(total, 2))

print("mean:", round(total / len(readings), 3))
````

## 5 · Three pieces for "close enough"

| piece | what it does |
|---|---|
| `abs(a - b)` | how far apart two numbers are — always positive |
| `a < b` | a comparison; gives back `True` or `False` (a `bool`) |
| `if test:` + `break` | run the indented line only when the test is `True`; `break` leaves the loop immediately |

````python
epsilon = 0.01

print(abs(5.5 ** 2 - 25))
print(abs(4.5 ** 2 - 25))
print(abs(5.5 ** 2 - 25) < epsilon)
print(abs(5.0 ** 2 - 25) < epsilon)
````

`break` in action: this loop *could* run 10 times, but stops as soon as `i * i` passes 20.

````python
for i in range(10):
    if i * i > 20:
        break
    print(i, "squared is", i * i)

print("stopped at i =", i)
````

## 7 · A peek at next session

A `while` loop repeats *as long as a condition is true*, so it can state the stopping rule directly — no safety cap, no `break`. Same algorithm, just run it:

````python
x = 25
epsilon = 0.01
guess = x / 2

while abs(guess * guess - x) >= epsilon:
    guess = (guess + x / guess) / 2

print(guess)
````
